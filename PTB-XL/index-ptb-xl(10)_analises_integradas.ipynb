# PTB-XL — Análise Exploratória

# 1. Base PTB-XL

## 1.1. Estrutura da Base de Dados

A base PTB-XL apresenta a seguinte organização de arquivos e diretórios:

    PTB-XL/
    ├── records100/
    ├── records500/
    ├── example_physionet.py
    ├── LICENSE.txt
    ├── ptbxl_database.csv
    ├── ptbxl_v102_changelog.txt
    ├── ptbxl_v103_changelog.txt
    ├── RECORDS
    ├── scp_statements.csv
    └── SHA256SUMS.txt

## 1.2. Descrição dos Arquivos

| Arquivo/Pasta | Descrição |
|---|---|
| `records100/` | Contém os sinais eletrocardiográficos amostrados a uma frequência de 100 Hz. |
| `records500/` | Contém os sinais eletrocardiográficos correspondentes, amostrados a uma frequência de 500 Hz. |
| `ptbxl_database.csv` | Arquivo principal de metadados, contendo informações como idade, sexo, identificadores dos pacientes, diagnósticos e caminhos dos arquivos dos sinais eletrocardiográficos. |
| `scp_statements.csv` | Contém as descrições dos códigos SCP-ECG e suas respectivas classificações e agrupamentos diagnósticos. |
| `example_physionet.py` | Código de exemplo desenvolvido em Python para demonstrar o carregamento, a leitura e a organização dos dados da PTB-XL. |
| `LICENSE.txt` | Contém as informações sobre a licença e as condições de utilização da base de dados. |
| `ptbxl_v102_changelog.txt` | Documento que registra as alterações realizadas na versão 1.0.2 da PTB-XL. |
| `ptbxl_v103_changelog.txt` | Documento que registra as alterações realizadas na versão 1.0.3 da PTB-XL. |
| `RECORDS` | Arquivo que apresenta a relação dos registros eletrocardiográficos disponibilizados na base. |
| `SHA256SUMS.txt` | Contém os códigos de verificação SHA-256 utilizados para conferir a integridade dos arquivos disponibilizados. |

# 1. Descrição das Variáveis da Base PTB-XL

O arquivo `ptbxl_database.csv` possui 28 colunas, que representam informações sobre os pacientes, exames eletrocardiográficos, diagnósticos, qualidade dos sinais e organização dos registros.

## 1.1. Tradução das 28 Colunas da PTB-XL

| Nº | Coluna original | Tradução | Significado |
|---|---|---|---|
| 1 | `ecg_id` | Identificador do ECG | Código único que identifica cada registro eletrocardiográfico. |
| 2 | `patient_id` | Identificador do paciente | Código que identifica o paciente associado ao exame. Um paciente pode possuir mais de um ECG. |
| 3 | `age` | Idade | Idade do paciente, em anos. |
| 4 | `sex` | Sexo | Sexo registrado: 0 = masculino; 1 = feminino. |
| 5 | `height` | Altura | Altura do paciente, em centímetros. |
| 6 | `weight` | Peso | Peso do paciente, em quilogramas. |
| 7 | `nurse` | Profissional responsável | Identificador do profissional de enfermagem associado ao registro. |
| 8 | `site` | Local de aquisição | Identificador do local onde o ECG foi registrado. |
| 9 | `device` | Equipamento | Dispositivo utilizado para registrar o ECG. |
| 10 | `recording_date` | Data do registro | Data e horário da aquisição do ECG. |
| 11 | `report` | Laudo | Texto contendo a interpretação ou descrição do ECG. |
| 12 | `scp_codes` | Códigos SCP-ECG | Dicionário contendo os códigos das alterações eletrocardiográficas e seus valores de confiança associados. |
| 13 | `heart_axis` | Eixo cardíaco | Informação sobre o eixo elétrico do coração. |
| 14 | `infarction_stadium1` | Estágio do infarto 1 | Primeira informação registrada sobre o estágio do infarto do miocárdio. |
| 15 | `infarction_stadium2` | Estágio do infarto 2 | Segunda informação registrada sobre o estágio do infarto do miocárdio. |
| 16 | `validated_by` | Validado por | Identificador do responsável pela validação do registro ou laudo. |
| 17 | `second_opinion` | Segunda opinião | Indica se houve uma segunda avaliação do ECG. |
| 18 | `initial_autogenerated_report` | Laudo inicial gerado automaticamente | Indica se o laudo inicial foi produzido automaticamente. |
| 19 | `validated_by_human` | Validado por humano | Indica se houve validação humana da interpretação. |
| 20 | `baseline_drift` | Desvio da linha de base | Informação sobre oscilações ou deslocamentos da linha de base do sinal. |
| 21 | `static_noise` | Ruído estático | Informação sobre presença de ruído estático no ECG. |
| 22 | `burst_noise` | Ruído em rajadas | Informação sobre interferências ou ruídos de curta duração. |
| 23 | `electrodes_problems` | Problemas nos eletrodos | Informação sobre problemas relacionados aos eletrodos utilizados na aquisição. |
| 24 | `extra_beats` | Batimentos extras | Informação sobre batimentos adicionais identificados no registro. |
| 25 | `pacemaker` | Marcapasso | Informação sobre presença ou atividade de marcapasso no ECG. |
| 26 | `strat_fold` | Grupo de particionamento estratificado | Identifica o fold de 1 a 10 utilizado na divisão dos dados. |
| 27 | `filename_lr` | Caminho do arquivo de baixa resolução | Localização do sinal digital amostrado a 100 Hz. |
| 28 | `filename_hr` | Caminho do arquivo de alta resolução | Localização do sinal digital amostrado a 500 Hz. |

## 1.2. Organização das Variáveis por Grupos

Para a análise exploratória da PTB-XL, as 28 variáveis podem ser organizadas em seis grupos, considerando suas características e finalidades.

| Grupo | Colunas |
|---|---|
| **1. Identificação** | `ecg_id`, `patient_id` |
| **2. Características dos pacientes** | `age`, `sex`, `height`, `weight` |
| **3. Aquisição do ECG** | `nurse`, `site`, `device`, `recording_date` |
| **4. Informações diagnósticas** | `report`, `scp_codes`, `heart_axis`, `infarction_stadium1`, `infarction_stadium2` |
| **5. Validação e qualidade dos registros** | `validated_by`, `second_opinion`, `initial_autogenerated_report`, `validated_by_human`, `baseline_drift`, `static_noise`, `burst_noise`, `electrodes_problems`, `extra_beats`, `pacemaker` |
| **6. Organização dos dados** | `strat_fold`, `filename_lr`, `filename_hr` |

# 2. Descrição das Variáveis do SCP Statements

O arquivo `scp_statements.csv` possui 13 colunas, responsáveis pela identificação, descrição, classificação e padronização dos códigos SCP-ECG utilizados na base PTB-XL.

## 2.1. Tradução das 13 Colunas

| Nº | Coluna original | Tradução | Significado |
|---|---|---|---|
| 1 | `Unnamed: 0` | `Codigo_SCP_ECG` | Identificador da declaração eletrocardiográfica, como NORM, IMI ou AFIB. |
| 2 | `description` | `Descricao` | Descrição resumida da condição ou alteração eletrocardiográfica. |
| 3 | `diagnostic` | `Diagnostico` | Indica se o código representa uma declaração diagnóstica. |
| 4 | `form` | `Morfologia` | Indica se o código está associado a características da forma do traçado eletrocardiográfico. |
| 5 | `rhythm` | `Ritmo` | Indica se o código está relacionado ao ritmo cardíaco. |
| 6 | `diagnostic_class` | `Classe_Diagnostica` | Categoria diagnóstica principal à qual o código pertence. |
| 7 | `diagnostic_subclass` | `Subclasse_Diagnostica` | Subdivisão mais específica da classe diagnóstica. |
| 8 | `Statement Category` | `Categoria_Declaracao` | Categoria atribuída à declaração na nomenclatura SCP-ECG. |
| 9 | `SCP-ECG Statement Description` | `Descricao_Declaracao_SCP_ECG` | Descrição padronizada associada ao código SCP-ECG. |
| 10 | `AHA code` | `Codigo_AHA` | Código correspondente à nomenclatura da American Heart Association. |
| 11 | `aECG REFID` | `Identificador_Referencia_aECG` | Identificador utilizado no padrão de ECG anotado (*annotated ECG*). |
| 12 | `CDISC Code` | `Codigo_CDISC` | Código associado aos padrões do Clinical Data Interchange Standards Consortium. |
| 13 | `DICOM Code` | `Codigo_DICOM` | Código correspondente à terminologia DICOM. |

## 2.2. Organização das Variáveis por Grupos

As 13 variáveis do arquivo `scp_statements.csv` podem ser organizadas em cinco grupos, considerando suas características e finalidades.

| Grupo | Colunas |
|---|---|
| **1. Identificação** | `Codigo_SCP_ECG` |
| **2. Descrição** | `Descricao`, `Descricao_Declaracao_SCP_ECG` |
| **3. Tipo de declaração** | `Diagnostico`, `Morfologia`, `Ritmo`, `Categoria_Declaracao` |
| **4. Classificação diagnóstica** | `Classe_Diagnostica`, `Subclasse_Diagnostica` |
| **5. Padronização terminológica** | `Codigo_AHA`, `Identificador_Referencia_aECG`, `Codigo_CDISC`, `Codigo_DICOM` |

### Distribuição da PTB-XL

```python
# Utilizar todos os registros
df_ptbxl_distribuicao = df_ptbxl.copy()

# Identificar a pasta de origem
df_ptbxl_distribuicao["pasta_origem"] = (
    df_ptbxl_distribuicao["filename_hr"]
    .str.split("/")
    .str[1]
)

# Identificar os arquivos
df_ptbxl_distribuicao["arquivo_dat"] = (
    df_ptbxl_distribuicao["filename_hr"] + ".dat"
)

df_ptbxl_distribuicao["arquivo_hea"] = (
    df_ptbxl_distribuicao["filename_hr"] + ".hea"
)

# Informações gerais
print("Total de registros:", len(df_ptbxl_distribuicao))
print("Total de pacientes:", df_ptbxl_distribuicao["patient_id"].nunique())
print("Total de pastas:", df_ptbxl_distribuicao["pasta_origem"].nunique())

# Visualizar somente os 5 primeiros registros
display(
    df_ptbxl_distribuicao[
        [
            "ecg_id",
            "patient_id",
            "pasta_origem",
            "arquivo_dat",
            "arquivo_hea",
            "age",
            "sex",
            "scp_codes"
        ]
    ].head(5)
)

# Distribuição dos registros por pasta
distribuicao_pastas = (
    df_ptbxl_distribuicao["pasta_origem"]
    .value_counts()
    .sort_index()
    .rename_axis("Pasta")
    .reset_index(name="Quantidade_ECGs")
)

display(distribuicao_pastas)
```

```python
import ast
import pandas as pd
from IPython.display import display, Markdown

# 1. Extrair os códigos presentes em cada ECG
codigos = df_ptbxl_pt["Codigos_SCP_ECG"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)

# 2. Contar quantos registros possuem cada código
frequencias = (
    codigos
    .apply(lambda x: list(x.keys()))
    .explode()
    .value_counts()
)

# 3. Preparar o dicionário SCP
df_guia = df_scp_statements_pt.copy()

# 4. Adicionar a frequência de cada código
df_guia["Quantidade_ECGs"] = (
    df_guia["Codigo_SCP_ECG"]
    .map(frequencias)
    .fillna(0)
    .astype(int)
)

# 5. Selecionar as informações
colunas = [
    "Codigo_SCP_ECG",
    "Descricao",
    "Diagnostico",
    "Morfologia",
    "Ritmo",
    "Classe_Diagnostica",
    "Subclasse_Diagnostica",
    "Categoria_Declaracao",
    "Descricao_Declaracao_SCP_ECG",
    "Codigo_AHA",
    "Identificador_Referencia_aECG",
    "Codigo_CDISC",
    "Codigo_DICOM",
    "Quantidade_ECGs"
]

df_guia = df_guia[colunas]

# 6. Organizar por frequência
df_guia = df_guia.sort_values(
    "Quantidade_ECGs",
    ascending=False
)

# 7. Reorganizar o índice começando em 1
df_guia = df_guia.reset_index(drop=True)
df_guia.index = range(1, len(df_guia) + 1)
df_guia.index.name = "Nº"

# 8. Exibir informações gerais
print("Total de códigos SCP-ECG:", len(df_guia))
print("Total de registros de ECG:", len(df_ptbxl_pt))
print("Códigos utilizados:", (df_guia["Quantidade_ECGs"] > 0).sum())
print("Códigos não utilizados:", (df_guia["Quantidade_ECGs"] == 0).sum())

# 9. Exibir todas as informações com índice
display(
    df_guia.style
    .set_properties(**{
        "text-align": "left",
        "white-space": "normal"
    })
)
```

Apenas registros após os três filtros acumulados. Não há novas remoções nem comparação antes/depois. Cada ECG pode possuir múltiplas declarações SCP-ECG; os gráficos apresentam nomes traduzidos.

```python
import pandas as pd
import numpy as np
import ast
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from IPython.display import display
sns.set_theme(style="whitegrid")
FILTROS = ["Problemas_Eletrodos", "Ruido_Estatico", "Marcapasso"]
df_eda = df_ptbxl_pt.loc[df_ptbxl_pt[FILTROS].isna().all(axis=1)].copy()
assert df_eda[FILTROS].isna().all().all()
assert df_eda["Identificador_ECG"].is_unique
mapa = df_descricoes_pt.drop_duplicates("Codigo_SCP_ECG").set_index("Codigo_SCP_ECG")["Descricao"].to_dict()
def codigos_scp(v):
    if isinstance(v,str):
        try: v=ast.literal_eval(v)
        except (SyntaxError,ValueError): return []
    return list(v) if isinstance(v,dict) else []
df_eda["Lista_Codigos"] = df_eda["Codigos_SCP_ECG"].apply(codigos_scp)
df_condicoes = df_eda.explode("Lista_Codigos").rename(columns={"Lista_Codigos":"Codigo"}).dropna(subset=["Codigo"]).copy()
df_condicoes["Descricao"] = df_condicoes["Codigo"].map(mapa)
sem_traducao = sorted(df_condicoes.loc[df_condicoes["Descricao"].isna(),"Codigo"].unique())
print("Códigos sem tradução:",sem_traducao)
df_condicoes["Descricao"] = df_condicoes["Descricao"].fillna("Descrição indisponível")
rotulos = df_condicoes[["Codigo","Descricao"]].drop_duplicates().copy()
dup = rotulos["Descricao"].duplicated(keep=False)
rotulos["Condicao"] = rotulos["Descricao"]
rotulos.loc[dup,"Condicao"] = rotulos.loc[dup,"Descricao"] + " (" + rotulos.loc[dup,"Codigo"] + ")"
df_condicoes = df_condicoes.merge(rotulos[["Codigo","Condicao"]],on="Codigo",how="left")
print("ECGs:",len(df_eda),"| Pacientes:",df_eda["Identificador_Paciente"].nunique())
display(df_condicoes[["Identificador_ECG","Idade","Sexo","Condicao"]].head(10))
```

## 1. Estatística descritiva — idade, altura e peso

```python
NUM = ["Idade","Altura_cm","Peso_kg"]
linhas=[]
for c in NUM:
    x=pd.to_numeric(df_eda[c],errors="coerce")
    linhas.append(dict(Feature=c,Validos=x.notna().sum(),Ausentes=x.isna().sum(),Media=x.mean(),Mediana=x.median(),Moda=x.mode().iloc[0] if not x.mode().empty else np.nan,Desvio_Padrao=x.std(),Minimo=x.min(),Q1=x.quantile(.25),Q3=x.quantile(.75),Maximo=x.max(),IIQ=x.quantile(.75)-x.quantile(.25)))
df_estatisticas=pd.DataFrame(linhas)
display(df_estatisticas.round(2))
df_ausentes=pd.DataFrame({"Feature":df_eda.columns,"Ausentes":df_eda.isna().sum().values})
df_ausentes["Percentual"]=df_ausentes["Ausentes"]/len(df_eda)*100
display(df_ausentes.round(2))
```

## 2. Histogramas, densidades e boxplots

```python
for COLUNA in NUM:
    x=pd.to_numeric(df_eda[COLUNA],errors="coerce").dropna()
    if x.empty: continue
    plt.figure(figsize=(12,5));sns.histplot(x,bins=30,kde=x.nunique()>1)
    plt.title("Distribuição de "+COLUNA);plt.xlabel(COLUNA);plt.ylabel("Quantidade de ECGs");plt.tight_layout();plt.show()
    plt.figure(figsize=(12,3));sns.boxplot(x=x);plt.title("Boxplot — "+COLUNA);plt.tight_layout();plt.show()
```

## 3. Variáveis categóricas — frequências e percentuais

```python
for COLUNA in ["Sexo","Eixo_Cardiaco","Validado_Por_Humano","Segunda_Opiniao","Grupo_Estratificado"]:
    s=df_eda[COLUNA]
    if COLUNA=="Sexo": s=s.map({0:"Masculino",1:"Feminino"})
    tab=s.fillna("Não informado").astype(str).value_counts().rename_axis("Categoria").reset_index(name="ECGs")
    tab["Percentual"]=tab["ECGs"]/len(df_eda)*100
    print(COLUNA);display(tab.round(2))
    plt.figure(figsize=(max(9,min(18,len(tab)*.8)),5));sns.barplot(data=tab,x="Categoria",y="ECGs",color="steelblue")
    plt.title("Distribuição de "+COLUNA);plt.xticks(rotation=45,ha="right");plt.tight_layout();plt.show()
```

## 4. Todas as declarações SCP-ECG — descrições traduzidas

```python
df_freq=(df_condicoes.groupby(["Codigo","Condicao"])["Identificador_ECG"].nunique().reset_index(name="ECGs").sort_values("ECGs",ascending=False))
df_freq["Percentual"]=df_freq["ECGs"]/len(df_eda)*100
display(df_freq[["Condicao","ECGs","Percentual"]].round(2))
plt.figure(figsize=(18,max(12,len(df_freq)*.37)));sns.barplot(data=df_freq,y="Condicao",x="ECGs",color="steelblue")
plt.title("Distribuição de todas as declarações SCP-ECG");plt.tight_layout();plt.show()
```

## 5. Vinte declarações mais frequentes

```python
top20=df_freq.head(20)
plt.figure(figsize=(16,10));ax=sns.barplot(data=top20,y="Condicao",x="ECGs",color="steelblue")
for b in ax.patches:
    ax.annotate(str(int(b.get_width())),(b.get_width(),b.get_y()+b.get_height()/2),xytext=(5,0),textcoords="offset points",va="center")
plt.xlim(0,top20["ECGs"].max()*1.15);plt.title("20 declarações mais frequentes");plt.tight_layout();plt.show()
```

## 6. Idade, altura e peso por declaração SCP-ECG

```python
top5_codigos=df_freq.head(5)["Codigo"].tolist()
df_top5=df_condicoes.loc[df_condicoes["Codigo"].isin(top5_codigos)].copy()
estatisticas_por_condicao=df_condicoes.groupby(["Codigo","Condicao"])[NUM].agg(["count","mean","median","std","min","max"])
display(estatisticas_por_condicao.round(2))
for COLUNA in NUM:
    plt.figure(figsize=(15,6));sns.boxplot(data=df_top5,x="Condicao",y=COLUNA)
    plt.title(COLUNA+" por declaração SCP-ECG");plt.xticks(rotation=25,ha="right");plt.tight_layout();plt.show()
```

## 7. Histograma e violin plot — idade por condição

```python
plt.figure(figsize=(14,6));sns.histplot(data=df_top5,x="Idade",hue="Condicao",bins=30,element="step",stat="density",common_norm=False)
plt.title("Distribuição de idade por declaração");plt.tight_layout();plt.show()
plt.figure(figsize=(15,7));sns.violinplot(data=df_top5,x="Condicao",y="Idade",inner="quartile",cut=0)
plt.title("Distribuição de idade por declaração");plt.xticks(rotation=25,ha="right");plt.tight_layout();plt.show()
```

## 8. Sexo por declaração SCP-ECG

```python
df_top5["Sexo_Descricao"]=df_top5["Sexo"].map({0:"Masculino",1:"Feminino"}).fillna("Não informado")
tab_sexo=df_top5.groupby(["Condicao","Sexo_Descricao"])["Identificador_ECG"].nunique().reset_index(name="ECGs")
display(tab_sexo)
plt.figure(figsize=(16,7));sns.barplot(data=tab_sexo,x="Condicao",y="ECGs",hue="Sexo_Descricao")
plt.xticks(rotation=25,ha="right");plt.title("Sexo por declaração SCP-ECG");plt.tight_layout();plt.show()
```

## 9. Eixo cardíaco e estágios de infarto por condição

```python
for COLUNA in ["Eixo_Cardiaco","Estagio_Infarto_1","Estagio_Infarto_2"]:
    tab=pd.crosstab(df_top5["Condicao"],df_top5[COLUNA].fillna("Não informado").astype(str))
    print(COLUNA);display(tab)
    plt.figure(figsize=(max(12,min(24,tab.shape[1]*.8)),6));sns.heatmap(tab,cmap="Blues",annot=tab.size<=100,fmt="d")
    plt.title(COLUNA+" por declaração");plt.tight_layout();plt.show()
```

## 10. Condições por grupo estratificado

```python
tab_fold=pd.crosstab(df_top5["Condicao"],df_top5["Grupo_Estratificado"])
display(tab_fold)
plt.figure(figsize=(13,6));sns.heatmap(tab_fold,annot=True,fmt="d",cmap="Blues")
plt.title("Declarações por grupo estratificado");plt.tight_layout();plt.show()
```

## 11. Correlações de Pearson e Spearman

```python
for metodo in ["pearson","spearman"]:
    matriz=df_eda[NUM].apply(pd.to_numeric,errors="coerce").corr(method=metodo)
    print(metodo);display(matriz.round(3))
    plt.figure(figsize=(7,5));sns.heatmap(matriz,annot=True,fmt=".2f",cmap="coolwarm",vmin=-1,vmax=1)
    plt.title("Correlação de "+metodo.title());plt.tight_layout();plt.show()
```

## 12. Dispersão — altura e peso por declaração

```python
plt.figure(figsize=(13,7));sns.scatterplot(data=df_top5,x="Altura_cm",y="Peso_kg",hue="Condicao",alpha=.45)
plt.title("Altura × peso por declaração SCP-ECG");plt.tight_layout();plt.show()
```

## 13. Cinco superclasses diagnósticas traduzidas

```python
guia_diag=df_scp_statements_pt.loc[pd.to_numeric(df_scp_statements_pt["Diagnostico"],errors="coerce")==1]
mapa_classes=guia_diag.dropna(subset=["Codigo_SCP_ECG","Classe_Diagnostica"]).set_index("Codigo_SCP_ECG")["Classe_Diagnostica"].to_dict()
nomes_classes={"NORM":"ECG normal","MI":"Infarto do miocárdio","STTC":"Alterações do segmento ST e da onda T","CD":"Distúrbios de condução","HYP":"Hipertrofia"}
def classes_de_ecg(codigos):
    return {mapa_classes[c] for c in codigos if c in mapa_classes and mapa_classes[c] in nomes_classes}
series_classes=df_eda["Lista_Codigos"].apply(classes_de_ecg)
df_superclasses=pd.DataFrame([{"Condicao":nome,"ECGs":sum(c in x for x in series_classes)} for c,nome in nomes_classes.items()])
df_superclasses["Percentual"]=df_superclasses["ECGs"]/len(df_eda)*100
display(df_superclasses.round(2))
plt.figure(figsize=(13,6));sns.barplot(data=df_superclasses,x="Condicao",y="ECGs",color="steelblue")
plt.xticks(rotation=20,ha="right");plt.title("Superclasses diagnósticas");plt.tight_layout();plt.show()
```

## 14. Número de declarações SCP-ECG por exame

```python
quantidade=df_eda["Lista_Codigos"].apply(len)
tab_quantidade=quantidade.value_counts().sort_index().rename_axis("Anotacoes_por_ECG").reset_index(name="ECGs")
display(tab_quantidade)
plt.figure(figsize=(12,5));sns.barplot(data=tab_quantidade,x="Anotacoes_por_ECG",y="ECGs",color="steelblue")
plt.title("Número de declarações SCP-ECG por exame");plt.tight_layout();plt.show()
print("Média:",round(quantidade.mean(),2),"Mediana:",quantidade.median(),"Máximo:",quantidade.max())
```
